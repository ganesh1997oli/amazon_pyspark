```python
# Create the project container in Unity Catalog.
spark.sql("""
    CREATE SCHEMA IF NOT EXISTS workspace.amazon_pyspark_backup
    """
)

# Create storage for source files and experiment outputs.
spark.sql("""
    CREATE VOLUME IF NOT EXISTS workspace.amazon_pyspark_backup.raw_files
""")

spark.sql("""
    CREATE VOLUME IF NOT EXISTS workspace.amazon_pyspark_backup.experiments
""")
```

```python
# Verify the storage objects
spark.sql("""
          SHOW VOLUMES IN workspace.amazon_pyspark_backup
""").show(truncate=False)
```
